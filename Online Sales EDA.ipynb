Filters the dataset to show all duplicate rows based on a subset of columns:
Invoice,InvoiceDate,Customer ID

Using keep=False ensures all copies of duplicates are returned, not just one.This helps detect:
-Transactions logged multiple times.
-Data entry errors.
-Potential issues in merging or importing datasets.

In retail datasets, it’s common for one invoice to contain multiple products.
In that case, Invoice, InvoiceDate, and Customer ID may be the same, but Description differs.
These are valid multi‑line invoices, not errors.
True duplicates are rows where all fields match (including product, quantity, and price).

```python
df_duplicated=df[df.duplicated(subset=["Invoice","InvoiceDate","Customer ID"],keep=False)]
```

```python
df_duplicated
```

We will create a separate sales dataframe 

```python
sales_df=df[df["Quantity"]>0].copy()
```

```python
sales_df.head()
```

AOV should be calculated using sales_df, which includes only completed sales and excludes cancelled or returned orders. This ensures the metric accurately reflects average sales performance without being reduced by negative transactions. Since AOV is a sales KPI, it should be calculated for all completed orders, regardless of whether CustomerID is available.
Average Order Value (AOV): The average revenue generated from each completed order, calculated by averaging the total revenue of all completed invoices.

```python
Average_Order_Value=sales_df.groupby("Invoice")["Revenue"].sum().mean().round(2)
```

```python
Average_Order_Value
```

Created a separate DataFrame containing only return/cancellation transactions.
To ensure you're capturing only genuine returns, you can combine both conditions that the row is selected only if both conditions are true:
Quantity is negative AND
Invoice starts with "C"

```python
returns_df = df[
    (df["Quantity"] < 0) &
    (df["Invoice"].astype(str).str.startswith("C"))
].copy()
```
